```python
import pandas as pd
import numpy as np
```

```python
airbnb=pd.read_csv('airbnb.csv', sep=',')
airbnb
```

```python
airbnb['last_review'] = pd.to_datetime(airbnb['last_review'], dayfirst=True, errors='coerce')
airbnb['first_review'] = pd.to_datetime(airbnb['first_review'], dayfirst=True, errors='coerce')
airbnb['host_since'] = pd.to_datetime(airbnb['host_since'], dayfirst=True, errors='coerce')
```

```python
airbnb.rename(columns={'log_price': 'price'}, inplace=True)
```

```python
airbnb['price'] = airbnb['price'].replace(r'[\$,]', '', regex=True)
airbnb['host_response_rate'] = airbnb['host_response_rate'].replace(r'[ %]', '', regex=True)
```

```python
airbnb['price'] = pd.to_numeric(airbnb['price'], errors='coerce')
airbnb['host_response_rate'] = pd.to_numeric(airbnb['host_response_rate'], errors='coerce')
```

```python
airbnb['review_scores_rating'] = airbnb['review_scores_rating'].replace(r'[\$,]', '', regex=True)
airbnb['review_scores_rating'] = pd.to_numeric(airbnb['review_scores_rating'], errors='coerce')
```

```python
airbnb['price'] = np.expm1(airbnb['price']).round(2)
```

```python
airbnb['beds'] = pd.to_numeric(airbnb['beds'], errors='coerce')
```

```python
airbnb['beds'] = airbnb['beds'].fillna(1)
```

```python
airbnb['bedrooms'] = pd.to_numeric(airbnb['bedrooms'], errors='coerce')
```

```python
airbnb['bedrooms'] = airbnb['bedrooms'].fillna(1)
```

```python
airbnb['bathrooms'] = pd.to_numeric(airbnb['bathrooms'], errors='coerce')
```

```python
airbnb['bathrooms'] = airbnb['bathrooms'].fillna(1)
```

```python
airbnb['instant_bookable'] = airbnb['instant_bookable'].str.lower() == 't'
```

```python
airbnb['host_has_profile_pic'] = airbnb['host_has_profile_pic'].str.lower() == 't'
```

```python
airbnb['host_identity_verified'] = airbnb['host_identity_verified'].str.lower() == 't'
```

```python
airbnb['host_response_rate'] = airbnb['host_response_rate'].fillna(0)
```

```python
airbnb['review_scores_rating'] = airbnb['review_scores_rating'].fillna(0)
```

```python
airbnb['zipcode'].hasnans
```

```python
airbnb['zipcode'] = airbnb['zipcode'].fillna(0)
```

```python
airbnb['zipcode'] = airbnb['zipcode'].astype(str).str.strip().str[:20]
```

```python
cols = ['host_has_profile_pic', 
        'host_identity_verified',
        'cleaning_fee',
        'instant_bookable']

for col in cols:
    airbnb[col] = airbnb[col].astype(str).str.lower().map({
        't': 1,
        'f': 0,
        'true': 1,
        'false': 0,
        'TRUE' : 1,
        'FALSE' : 0,
        '1' : 1,
        '0' : 0
    })
```
